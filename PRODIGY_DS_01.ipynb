```python

import pandas as pd
import matplotlib.pyplot as plt

data = {
    'Age': [
        18, 19, 20, 21, 22,
        20, 21, 22, 23, 24,
        19, 20, 21, 22, 25,
        23, 24, 25, 26, 27,
        28, 29, 30, 22, 21
    ]
}

df = pd.DataFrame(data)

print("Student Age Dataset:")
print(df)

plt.figure(figsize=(8, 5))

plt.hist(
    df['Age'],
    bins=6,
    edgecolor='black'
)

plt.title('Distribution of Student Ages')
plt.xlabel('Age')
plt.ylabel('Number of Students')

plt.grid(axis='y', alpha=0.3)

plt.show()

plt.savefig('student_age_distribution.png', dpi=300, bbox_inches='tight')
```

### Conclusion

In this project, I created a histogram using Python, Pandas, and Matplotlib to visualize the distribution of student ages. The histogram helps us understand how students are distributed across different age groups. It provides a simple visual representation of the dataset and makes it easier to identify common age ranges and patterns.
